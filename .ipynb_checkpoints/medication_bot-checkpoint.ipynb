```python
import pandas as pd
from llama_index.core.indices.vector_store.base import VectorStoreIndex
```

```python
from llama_index.core import Document
```

```python
from langchain.chat_models import ChatOpenAI
```

```python
# Loading and merging the sets (incl. "date" info)

df_train = pd.read_csv("data/drugsComTrain_raw.tsv", sep='\t', index_col=[0])
df_test = pd.read_csv("data/drugsComTest_raw.tsv", sep='\t', index_col=[0])
```

```python
df = pd.concat([df_train, df_test]).reset_index(drop=True)
```

```python
# Cleaning dups

df = df.drop_duplicates(subset=["condition" ,"review", "rating"]).reset_index(drop=True)
```

```python
df.rating = df.rating.astype('int')
```

```python
df
```

```python
# Creating documents for each review

documents = [
    Document(text=f"Drug: {row['drugName']}, Condition: {row['condition']}, Review: {row['review']}, Rating: {row['rating']}, Date: {row['date']}")
    for _, row in df.iterrows()
]
```

```python
# Building the index from documents

index = VectorStoreIndex.from_documents(documents)
```

```python
# 4. Set up the LLM
llm = ChatOpenAI(model_name="gpt-4", openai_api_key="YOUR_OPENAI_API_KEY")
```

```python
# Function to ask questions about a drug

def ask_about_drug(drugName, question):
    # Search for relevant reviews
    response = index.query(f"Find reviews about {drugName} relevant to: {question}")
    
    # Combine LLM’s knowledge with review data
    answer = llm.predict(f"Based on these reviews: {response}, answer this question: {question}")
    return answer

# Example usage
if __name__ == "__main__":
    drug = "Aspirin"
    question = "What are the most common side effects?"
    print(ask_about_drug(drug, question))
```
